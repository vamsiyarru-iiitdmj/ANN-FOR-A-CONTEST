```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
files = ['hack1 (2).csv',
'hack10.csv',
'hack100.csv',
'submission3.csv',
'hack2 (1).csv',]
df1 = pd.DataFrame()
count = 0
for i in files:
    
    df = pd.read_csv(i)
    if count == 0:
        df1['row_id'] = df['row_id']
        df1['Diabetes_binary'] = df['Diabetes_binary']
    df1['Diabetes_binary'] += df['Diabetes_binary']

    count += 1
# df = pd.read_csv("submission_v6 (1).csv")
# df1['Diabetes_binary'] += 2*(df['Diabetes_binary'])

# df1["Diabetes_binary"] /= (len(files) + 2)

def condition(a):
    if a >= 0.90:
        return 1
    else:
        return 0

df1['Diabetes_binary'] = df1["Diabetes_binary"].apply(condition)

df1.to_csv('combined1.csv', index = False)
display(df1.describe())
sns.kdeplot(data = df1, x = 'row_id', hue = 'Diabetes_binary')
plt.show()
plt.close()
df1['Diabetes_binary'].value_counts(normalize= True)

```
